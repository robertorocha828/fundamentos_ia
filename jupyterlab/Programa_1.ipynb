### Modificacion de un array

```python
# Creacion de un array unidimensional inicializandpo con el rango de
array1 = np.arange(28)
print("Shape:", array1.shape)
print("Array 1:", array1)
```

```python
# Cambiar las dimenciones del array y sus longitudes
array1.shape = (7,4)
print("Shape:", array1.shape)
print("Array 1:\n", array1)
```

```python
# El ejemplo anterior devuelve un nuevo Array que apunta
# Importante: MOdificaciones es un Array, modificaron
array2 = array1.reshape(4,7)
print("Shape:", array2.shape)
print("Array 2:\n", array2)
```

```python
# Modificacion del nuevo array devuelto
array2[0,3] = 20
print("Array 2:\n",array2)
```

```python
#Devuelve el array, devolviendo un nuevo array 
#IMportante: le nuevo array 
print("Array 1:\n",array1.revel())
```
