```python
"""Calificación aprobatoria con contador de intentos
Escribe un programa que pida al usuario su nota (del 0 al 100).
    Si la nota ingresada no está entre 0 y 100, debe mostrar "Nota inválida" y sumar 1 a un contador de errores.
    Si la nota está en el rango válido y es mayor o igual a 60, debe mostrar "Aprobado". De lo contrario, debe mostrar "Reprobado".
    Al finalizar, debe mostrar el total de errores registrados (que será 0 o 1)."""

errores = 0
nota_usuario = int(input("Ingrese su nota (0 al 100): "))
if nota_usuario < 0 or nota_usuario > 100:
    errores +=1
    print("Nota invalida")
elif nota_usuario >= 60:
    print("Aprobado")
else:
    print("Reprobado")

print(f"El ttoal errores son: {errores}")
```

```python
"""Calcula el total a pagar de una compra según las siguientes reglas:
  Si la compra es menor o igual a $50, no hay descuento (0%).
  Si es mayor a $50 y menor o igual a $100, aplica un 10% de descuento.
  Si es mayor a $100, aplica un 20% de descuento."""

compra = float(input("Ingresa el monto de tu compra: "))
if compra <= 50:
  descuento = 0
elif compra <= 100:
  descuento = compra * 0.10
else:
  descuento = compra * 0.20

total_a_pagar = compra - descuento

print(f"Descuento aplicado: ${descuento}")
print(f"Total a pagar: ${total_a_pagar}")
```

```python
"""Una tienda aplica descuentos según el costo de compra: menos de $100, sin descuento; de $100 a $500, un 5 %; más de $500, un 10 %. 
Solicita el importe y muestra el total a pagar. Si el costo es negativo, informa que no es válido."""

costo = float(input("Ingrese el costo del producto: "))

if costo < 0:
    print("Precio no valido")
else:
    if costo < 100:
        print("Sin descuento")
    elif costo <= 500:
        descuento = costo*0.05
        print(f"Su descuento es de {descuento}")
    else:
        descuento = costo*0.1
        print(f"Su descuento es de {descuento}")
```

```python
#El docente ingresa 4 notas, calcular el promedio, Si el promedio es < a 35 imprimir "Reprobado", si es menor a 51 "Recuperacion" y si es mayor o iguala 51 "Aprobado
matematicas = float(input("Ingrese la nota de matematicas: "))
fisica = float(input("Ingrese la nota de fisica: "))
quimica = float(input("Ingrese la nota de quimica: "))

promedio = (matematicas+fisica+quimica)/3

if promedio < 35:
    print("Reprobado")
elif promedio < 51:
    print(f"El estudiante debe Recuperacion, su nota es de: {promedio}")
elif promedio >= 51:
    print(f"EL estudiante Aprobado con la nota de: {promedio}")
```

```python
"""Escribe un programa que simule la entrada a un evento con zona VIP y zona General, aplicando las siguientes reglas:
    Solicita al usuario su edad (entero) y el tipo de boleto que compró (ingresar "VIP" o "General").
    Validación de edad:
        Si la edad es menor a 18, debe mostrar "Acceso denegado: Evento solo para mayores de edad".
    Clasificación por boleto (si es mayor o igual a 18):
        Si el boleto es "VIP":
            Si la persona tiene 60 años o más, se le otorga un beneficio extra y muestra "Acceso a Zona VIP con pase directo al lounge".
            Si tiene menos de 60 años, muestra "Acceso a Zona VIP".
        Si el boleto es "General":
            Muestra "Acceso a Zona General".
        Si ingresa un tipo de boleto diferente a "VIP" o "General", muestra "Tipo de boleto no válido"."""

edad = int(input("Ingrese su edad: "))
boleto = input("Ingrese su tipo de boleto (VIP / General): ").strip()
if edad < 18:
    print("Acceso denegado: Evento solo para mayores de edad")
else:
    if boleto == "VIP" or boleto == "vip":
        if edad >= 60:
            print("Acceso a Zona VIP con pase directo al lounge")
        else:
            print("Acceso a Zona VIP")
    elif boleto == "General" or boleto == "general":
        print("Acceso a Zona General")
    else:
        print("Tipo de boleto no válido")
```

```python
"""Pide la edad y si ha tenido accidentes ("si" o "no"). Muestra el nivel de riesgo según estas reglas:
    Menor de 18: "No elegible para póliza de seguro"
    De 18 a 25 años:
        Con accidentes: "Riesgo Alto: Tarifa máxima"
        Sin accidentes: "Riesgo Medio: Tarifa estándar"
    Mayor de 25 años:
        Con accidentes: "Riesgo Medio: Tarifa estándar"
        Sin accidentes: "Riesgo Bajo: Aplica descuento preferencial"""

edad = int(input("Ingrese su edad: "))
if edad < 18:
    print("No elegible para póliza de seguro")
else:
    accidentes = input("¿Ha tenido accidentes? (si/no): ").lower().strip()
    if edad <= 25:
        if accidentes == "si":
            print("Riesgo Alto: Tarifa máxima")
        else:
            print("Riesgo Medio: Tarifa estándar")
    else:
        if accidentes == "si":
            print("Riesgo Medio: Tarifa estándar")
        else:
            print("Riesgo Bajo: Aplica descuento preferencial")
```
