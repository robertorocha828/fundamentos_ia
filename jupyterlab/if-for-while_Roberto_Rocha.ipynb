```python
"""Calcula el total a pagar de una compra según las siguientes reglas:
  Si la compra es menor o igual a $50, no hay descuento (0%).
  Si es mayor a $50 y menor o igual a $100, aplica un 10% de descuento.
  Si es mayor a $100, aplica un 20% de descuento."""

compra = float(input("Ingresa el monto de tu compra: "))
if compra <= 50:
  descuento = 0
elif compra <= 100:
  descuento = compra * 0.10
else:
  descuento = compra * 0.20

total_a_pagar = compra - descuento

print(f"Descuento aplicado: ${descuento}")
print(f"Total a pagar: ${total_a_pagar}")
```

### for

```python
"""Registro de compras en una caja registradora
Solicita el precio de 5 productos comprados y calcula el total de la compra."""

total_compra = 0
for i in range(1, 6):
    precio = float(input(f"Ingrese el precio del producto {i}: $"))
    total_compra += precio

print(f"Total a pagar en caja: ${total_compra:.2f}")
```

```python
"""Control de asistencia a una clase
Pide la asistencia de 6 estudiantes ("p" para presente, "f" para falta) y cuenta cuántos asistieron."""

asistencias = 0
for i in range(1, 7):
    estado = input(f"Estudiante {i} - ¿Presente o Falta? (p/f): ").lower()
    if estado == "p":
        asistencias += 1

print(f"Total de estudiantes presentes: {asistencias} de 6")
```

```python
"""Evaluación de satisfacción de clientes
Solicita la calificación del servicio (del 1 al 5) a 4 clientes y cuenta cuántos dieron una nota excelente (5)."""

excelentes = 0
for i in range(1, 5):
    calificacion = int(input(f"Cliente {i}, califique el servicio (1 al 5): "))
    if calificacion == 5:
        excelentes += 1

print(f"Clientes totalmente satisfechos: {excelentes}")
```

```python
"""Registro de consumo de combustible
Pide los galones cargados en 4 viajes y cuenta cuántos viajes requirieron más de 10 galones."""

viajes_altos = 0
total_galones = 0

for i in range(1, 5):
    galones = float(input(f"Ingrese galones consumidos en el viaje {i}: "))
    total_galones += galones
    if galones > 10:
        viajes_altos += 1

print(f"Total de galones consumidos: {total_galones:.2f}")
print(f"Viajes con consumo alto (>10 galones): {viajes_altos}")
```

```python
"""Control de gastos semanales con presupuesto
Solicita el gasto diario de lunes a viernes (5 días) y verifica cuántos días se superó el límite diario de $20."""

dias_excedidos = 0
gasto_total = 0

for i in range(1, 6):
    gasto = float(input(f"Ingrese el gasto del día {i}: $"))
    gasto_total += gasto
    if gasto > 20:
        dias_excedidos += 1

print(f"Gasto acumulado de la semana: ${gasto_total:.2f}")
print(f"Días que superaste el límite de $20: {dias_excedidos}")
```

```python
"""Registro de temperaturas para control climático
Pide la temperatura registrada a 4 horas distintas del día y cuenta cuántas lecturas superaron el umbral crítico de 28 °C."""

alertas = 0

for i in range(1, 5):
    temp = float(input(f"Ingrese la temperatura de la lectura {i} (°C): "))
    if temp > 28.0:
        alertas += 1

print(f"Lecturas con sobrecalentamiento: {alertas}")
```

```python
"""Control de depósitos de ahorro
Solicita el valor depositado durante 4 meses y calcula el total ahorrado acumulado."""

total_ahorrado = 0

for mes in range(1, 5):
    deposito = float(input(f"Ingrese el monto a depositar en el mes {mes}: $"))
    total_ahorrado += deposito

print(f"Fondo total ahorrado al cabo de 4 meses: ${total_ahorrado:.2f}")
```

```python
"""Sensor de velocidad en carretera
Solicita la velocidad de 5 vehículos que pasan por un punto de control y cuenta cuántos excedieron el límite de 90 km/h."""

infracciones = 0

for i in range(1, 6):
    velocidad = float(input(f"Velocidad capturada para el vehículo {i} (km/h): "))
    if velocidad > 90:
        infracciones += 1

print(f"Total de vehículos sancionados: {infracciones}")
```

```python
"""Promedio de notas y control de aprobadosPide la nota final (0 a 100) de 5 estudiantes, 
calcula el promedio del grupo y cuenta cuántos aprobaron ($\ge 60$)"""

suma_notas = 0
aprobados = 0

for i in range(1, 6):
    nota = float(input(f"Ingrese la nota del estudiante {i}: "))
    suma_notas += nota
    if nota >= 60:
        aprobados += 1

promedio = suma_notas / 5
print(f"Promedio general de la clase: {promedio:.2f}")
print(f"Cantidad de estudiantes aprobados: {aprobados}")
```

### While

```python
"""Cajero automático (Cierre de cuenta por saldo insuficiente)
Permite retirar dinero mientras haya saldo disponible."""

saldo = 200.0

while saldo > 0:
    retiro = float(input(f"Saldo disponible: ${saldo:.2f}. Ingrese monto a retirar: $"))
    if retiro <= 0:
        print("Monto no válido.")
    elif retiro <= saldo:
        saldo -= retiro
        print(f"Retiro exitoso. Saldo restante: ${saldo:.2f}")
    else:
        print("Saldo insuficiente para realizar ese retiro.")

print("Su cuenta se ha quedado sin saldo. Operación finalizada.")
```

```python
"""Calculadora de compras hasta ingresar 0
Registra el precio de productos uno a uno hasta que el usuario ingresa 0 para finalizar la factura."""

total_factura = 0.0
cant_productos = 0
precio = -1.0

while precio != 0:
    precio = float(input("Ingrese precio del producto (o 0 para pagar): $"))
    if precio > 0:
        total_factura += precio
        cant_productos += 1
    elif precio < 0:
        print("El precio no puede ser negativo.")

print(f"Productos registrados: {cant_productos}")
print(f"Total a pagar: ${total_factura:.2f}")
```

```python
"""Bloqueo de tarjeta por intentos de PIN incorrecto
Permite hasta 3 intentos fallidos para ingresar el PIN correcto de una tarjeta."""

pin_correcto = "1234"
intentos_fallidos = 0
acceso_concedido = False

while intentos_fallidos < 3 and not acceso_concedido:
    pin_ingresado = input("Ingrese su PIN de 4 dígitos: ")
    if pin_ingresado == pin_correcto:
        acceso_concedido = True
        print("Acceso concedido. Bienvenido a su banca en línea.")
    else:
        intentos_fallidos += 1
        print(f"PIN incorrecto. Intentos restantes: {3 - intentos_fallidos}")

if not acceso_concedido:
    print("Tarjeta bloqueada por seguridad. Contacte a su banco.")
```

```python
"""Meta de ahorro mensual
Solicita depósitos hasta alcanzar o superar una meta de ahorro predefinida ($500)."""

meta_ahorro = 500.0
acumulado = 0.0
depositos_realizados = 0

while acumulado < meta_ahorro:
    monto = float(input(f"Llevas ${acumulado:.2f} de ${meta_ahorro:.2f}. Ingrese depósito: $"))
    if monto > 0:
        acumulado += monto
        depositos_realizados += 1
    else:
        print("El depósito debe ser mayor a $0.")

print(f"¡Felicidades! Meta alcanzada en {depositos_realizados} depósitos.")
print(f"Total ahorrado: ${acumulado:.2f}")
```

```python
"""Control de tanque de agua por nivel
Rellena un tanque de agua pidiendo litros acumulados hasta alcanzar los 100 litros de capacidad."""

capacidad_maxima = 100.0
nivel_actual = 0.0
cargas = 0

while nivel_actual < capacidad_maxima:
    litros = float(input(f"Nivel actual: {nivel_actual:.1f}L / {capacidad_maxima}L. Litros a ingresar: "))
    if litros > 0:
        if nivel_actual + litros <= capacidad_maxima:
            nivel_actual += litros
            cargas += 1
        else:
            print(f"Alerta: Excede la capacidad. Solo puedes agregar {capacidad_maxima - nivel_actual:.1f}L más.")
    else:
        print("Ingrese una cantidad válida.")

print(f"Tanque lleno. Se completó en {cargas} recargas.")
```
